# Operácie s číslami v Pythone

V tomto notebooku sa naučíte pracovať s matematickými operáciami a objavíte, ako Python počíta s rôznymi typmi čísel.

## Základné aritmetické operácie

Python vie pracovať s číslami rovnako ako kalkulačka. Poďme si vyskúšať základné operácie.

### Sčítanie a odčítanie

```python
# Sčítanie
5 + 3
```

```python
# Odčítanie
10 - 4
```

```python
# Dajú sa použiť aj záporné čísla
-5 + 8
```

```python
# S premennými
a = 15
b = 7
vysledok = a + b
print(vysledok)
```

### Násobenie

```python
# Násobenie sa v Pythone značí hviezdičkou 
6 * 7
```

```python
# Násobenie viacerých čísel
2 * 3 * 4
```

```python
# Násobenie záporným číslom
5 * (-3)
```

### Delenie

V Pythone existujú dva typy delenia!

```python
# Bežné delenie /
10 / 3
```

```python
# Bežné delenie vždy vráti desatinné číslo (float)
8 / 2
```

Pozorovanie: Čo si všimnete pri výsledku 8 / 2? Prečo je výsledok 4.0 a nie 4?

```python
# Overme si typ výsledku
vysledok_1 = 10 / 3
vysledok_2 = 8 / 2

print(type(vysledok_1))
print(type(vysledok_2))
```

## Celočíselné delenie //

Ak chceme výsledok bez desatinnej časti, použijeme celočíselné delenie s operátorom //

```python
# Celočíselné delenie - vráti len celú časť
10 // 3
```

```python
# Ďalšie príklady
print(17 // 5)
print(20 // 6)
print(9 // 2)
```

```python
# Čo sa stane, keď delíme dve celé čísla?
vysledok = 9 // 5
print(vysledok)
print(type(vysledok))
```

```python
# Čo ak použijeme desatinné číslo?
vysledok_1 = 9.0 // 5
vysledok_2 = 9 // 5.0

print(vysledok_1)
print(vysledok_2)
print(type(vysledok_1))
print(type(vysledok_2))
```

Pravidlo: Ak je aspoň jedno číslo float, výsledok bude float!

## Modulo (zvyšok po delení) %

Operátor % (modulo) vráti zvyšok po delení.

```python
# 10 delené 3 je 3 so zvyškom 1
10 % 3
```

```python
# Keď je číslo deliteľné bezo zvyšku, výsledok je 0
10 % 2
```

```python
# Ďalšie príklady
print(17 % 5) # 17 = 35 + 2, zvyšok je 2
print(20 % 6) # 20 = 36 + 2, zvyšok je 2
print(15 % 4) # 15 = 34 + 3, zvyšok je 3
```

### Praktické použitie modulo

Modulo sa často používa na zistenie, či je číslo párne alebo nepárne.

```python
# Párne číslo: zvyšok po delení 2 je 0
print(8 % 2)
print(10 % 2)
```

```python
# Nepárne číslo: zvyšok po delení 2 je 1
print(7 % 2)
print(13 % 2)
```

Otázka: Ako by ste zistili, či je číslo deliteľné tromi?

## Umocňovanie 

Na umocňovanie (mocninu) používame operátor **

```python
# 2 na druhú
2 ** 2
```

```python
# 2 na tretiu
2 **3
```

```python
# 5 na štvrtú
5 **4
```

```python
# Dá sa to aj s premennými
zaklad = 3
exponent = 5
vysledok = zaklad ** exponent
print(vysledok)
```

### Druhá odmocnina

Pomocou mocniny 0.5 môžeme vypočítať druhú odmocninu:

```python
# Druhá odmocnina z 16
16 ** 0.5
```

```python
# Druhá odmocnina z 25
25 ** 0.5
```

## Poradie operácií

Python rešpektuje matematické poradie operácií: zátvorky, mocniny, násobenie/delenie, sčítanie/odčítanie

```python
# Násobenie má prednosť pred sčítaním
2 + 3 * 4
```

```python
# Pomocou zátvoriek zmeníme poradie
(2 + 3) * 4
```

```python
# Zložitejší príklad
10 + 2 * 5 ** 2
```

Otázka: Ako Python vypočítal výsledok vyššie? V akom poradí prebehli operácie?

```python
# So zátvorkami je to jasnejšie
10 + (2 * (5 ** 2))
```

Tip: Keď si nie ste istí poradím operácií, použite zátvorky pre lepšiu čitateľnosť!

## Typy čísel pri operáciách

Čo sa stane s dátovými typmi, keď kombinujeme int a float?

```python
# int + int = int
vysledok = 5 + 3
print(vysledok)
print(type(vysledok))
```

```python
# int + float = float
vysledok = 5 + 3.0
print(vysledok)
print(type(vysledok))
```

```python
# int * float = float
vysledok = 5 * 2.0
print(vysledok)
print(type(vysledok))
```

```python
# Delenie / vždy vráti float
vysledok = 10 / 2
print(vysledok)
print(type(vysledok))
```

Pravidlo: Keď kombinujeme int a float, výsledok je vždy float!

## Chyby pri počítaní

Nie všetky operácie sú v matematike povolené. Čo sa stane, keď skúsime niečo zakázané?

### Delenie nulou

```python
# Bežné delenie nulou
10 / 0
```

Chyba: ZeroDivisionError - Python nedovolí deliť nulou!

```python
# Celočíselné delenie nulou
10 // 0
```

```python
# Modulo nulou
10 % 0
```

## Skrátené operátory

Python má skrátené zápisy pre časté operácie s premennými.

```python
x = 5
print(x)
```

```python
x += 10
```

```python
print(x)
```

```python
# Klasický spôsob: zvýšiť hodnotu o 5
x = 10
x = x + 5
print(x)
```

```python
# Skrátený zápis: += 
x = 10
x += 5 # Rovnaké ako x = x + 5
print(x)
```

```python
# Ďalšie skrátené operátory
cislo = 100

cislo -= 20 # Rovnaké ako cislo = cislo - 20
print(cislo)

cislo *= 2 # Rovnaké ako cislo = cislo * 2
print(cislo)

cislo /= 4 # Rovnaké ako cislo = cislo / 4
print(cislo)
```

```python
# Funguje aj pre umocňovanie
zaklad = 2
zaklad = 5 # Rovnaké ako zaklad = zaklad 5
print(zaklad)
```
