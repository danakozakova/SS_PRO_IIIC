## A) Celé čísla, delenie a modulo

### `int` v Pythone nemá limit
V C++ má `int` pevný počet bitov a pri pretečení sa „zabalí". Python `int` je **neobmedzený** — spusti:


````python
print(2 ** 100)
print(2 ** 1000)
````

### Delenie: `/`, `//`, `%`, `divmod`
- `/` vždy vráti `float`
- `//` je **celočíselné delenie nadol** (floor)
- `%` je zvyšok, `divmod(a, b)` vráti oboje naraz


````python
print(7 / 2, 7 // 2, 7 % 2)
print(divmod(7, 2))
````

### ⚠️ Pozor: modulo so zápornými číslami (Python vs C++)
V Pythone má výsledok `%` **znamienko deliteľa**, a `//` zaokrúhľuje **nadol** (nie k nule ako C++).
Najprv **predpovedz** výstup, potom spusti:


````python
print(-7 % 3)      # C++ by dalo -1, Python?
print(-7 // 3)     # C++ by dalo -2, Python?
print(divmod(-7, 3))
````

**❓ Otázka:** Prečo `-7 % 3` v Pythone vyjde `2`? (Pomôcka: platí `a == (a // b) * b + (a % b)`.)

## B) Desatinné čísla a presnosť (IEEE-754)

`float` je v Pythone (aj v C++) 64-bitové číslo podľa normy IEEE-754 — a niektoré čísla sa doň **nezmestia presne**. Spusti:


````python
print(0.1 + 0.2)
print(0.1 + 0.2 == 0.3)
````

To nie je chyba Pythonu — `0.1` sa v dvojkovej sústave nedá zapísať presne. Ako sa brániť:
- porovnávaj cez `math.isclose(...)`, nie `==`
- alebo počítaj presne cez `Fraction` / `Decimal`


````python
import math
from fractions import Fraction
from decimal import Decimal

print(math.isclose(0.1 + 0.2, 0.3))
print(Fraction(1,10) + Fraction(2,10) == Fraction(3,10))
print(Decimal('0.1') + Decimal('0.2') == Decimal('0.3'))
````

## C) Sústavy, bitové operácie a formátovanie

### Prevody medzi sústavami


````python
print(bin(213), hex(213), oct(213))
print(int('11010101', 2))   # z dvojkovej
print(int('D5', 16))        # zo šestnástkovej
````

### Bitové operácie
`&` (and), `|` (or), `^` (xor), `~` (negácia), `<<` `>>` (posun). Spusti a rozmysli:


````python
print(5 & 3, 5 | 3, 5 ^ 3)
print(~5)          # prečo -6?
print(5 << 1, 5 >> 1)
````

### Formátovanie čísel v `f`-stringoch
Užitočné pri výpisoch: `:.2f` (2 desatinné), `:b`/`:x` (sústava), `:,` (oddeľovač tisícov).


````python
x = 255
print(f"{3.14159:.2f}")
print(f"{x:b}  {x:x}")
print(f"{1000000:,}")
````
