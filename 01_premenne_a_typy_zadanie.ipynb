## B) Typy a dynamické typovanie

V **C++** typ napíšeš dopredu a nezmeníš (`int a = 7;`). V Pythone to tak nie je — spusti a pozoruj:


```python
x = 7
print(type(x))
x = 'sedem'
print(type(x))   # to isté meno, iný typ!
```

Typ nesie **hodnota**, nie premenná. Premenná je len **menovka** nalepená na hodnote.


### Základné typy
`int`, `float`, `str`, `bool` (`True`/`False`), `None`. Over typ cez `type()` a `isinstance()`:


```python
print(type(58), type(99.9), type('abc'), type(True), type(None))
print(isinstance(58, int))
print(True + True)   # bool je podtyp int → 2
print(3 / 2, 3 // 2) # / je float, // je celočíselné
```

### Typické pasce
**Odkomentuj a spusti** — pozoruj chyby:


```python
# '5' + 5      # ⚠️ TypeError: str + int
# int('abc')   # ⚠️ ValueError
```

### Konverzie typov
`int()`, `float()`, `str()`, `bool()`. Vstup z `input()` je **vždy** `str`!


```python
print(int('12'), float('12.5'), str(58))
cislo = input('Zadaj číslo: ')
print(type(cislo))            # str!
print(int(cislo) + 1)         # najprv konverzia
```

## C) Typové anotácie (`type hints`)

Z C++ poznáte statické typy. V Pythone majú **dobrovoľnú** obdobu — poznámku pre čitateľa a nástroje. Python ich **nevynucuje**.


```python
vek: int = 18
meno: str = 'Jana'

def pozdrav(meno: str, krat: int) -> str:
    return (meno + ' ') * krat

print(pozdrav('Ahoj', 3))
```

Prečo ich chceme: **čitateľnosť**, IDE napovie, chyby sa odhalia skôr — a dobré typy sú aj lepšie zadanie pre AI aj pre kolegu. *Typy sú kontrakt.*
