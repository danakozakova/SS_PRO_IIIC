### Pridanie (registrácia) vlastného fontu

````python
from reportlab.pdfbase.ttfonts import TTFont

# Registrácia fontu Times New Roman
pdfmetrics.registerFont(TTFont('TimesNewRoman', 'times.ttf'))

# Kontrola, či bol font pridaný
pdfmetrics.getRegisteredFontNames()
````

**Ako použiť vlastné TTF fonty do PDF**

Dve možnosti:

**1. Použiť systémové fonty** – súbory .ttf, ktoré už máte v počítači (times.ttf, arial.ttf...)

**2. Stiahnuť TTF súbor z internetu:**
- Google Fonts (https://fonts.google.com) – zadarmo
- DaFont (https://www.dafont.com) – zadarmo
- Font Squirrel (https://www.fontsquirrel.com) – zadarmo

Po stiahnutí TTF súbor nahrajte do rovnakého priečinka ako notebook, alebo zadajte celú cestu:

```python
# Font v rovnakom priečinku ako notebook
pdfmetrics.registerFont(TTFont('MojFont', 'moj_font.ttf'))

# Font s celou cestou
pdfmetrics.registerFont(TTFont('MojFont', 'C:/Fonts/moj_font.ttf'))
```

`TTFont` hľadá fonty v aktuálnom (zadanom) priečinku a potom v predvolených priečinkoch:
- **Windows:** `C:\Windows\Fonts`
- **macOS:** `/Library/Fonts`
- **Unix:** `/usr/share/fonts`

> ### 💻 Poznámka pre študentov na macOS
>
> Bunky s `TTFont('TimesNewRoman', 'times.ttf')` fungujú na školských **Windows** PC, kde `reportlab` nájde `times.ttf` automaticky v `C:\Windows\Fonts`. Na **Macu** súbor `times.ttf` neexistuje, preto si vyber jednu z možností:
>
> **1. Použiť Times New Roman z macOS** (ak je nainštalovaný) – zadaj celú cestu:
> ```python
> pdfmetrics.registerFont(TTFont('TimesNewRoman',
>     '/System/Library/Fonts/Supplemental/Times New Roman.ttf'))
> ```
>
> **2. Nepoužiť vlastný font vôbec** (najjednoduchšie) – `reportlab` má zabudované fonty bez súboru: `Helvetica`, `Times-Roman`, `Courier`. Stačí ich priamo nastaviť a bunku s `registerFont` preskočiť:
> ```python
> c.setFont('Times-Roman', 35)   # namiesto 'TimesNewRoman'
> ```
>
> **3. Stiahnuť voľný font** – napr. *Tinos* alebo *Liberation Serif* (metricky zhodné s Times New Roman) z [Google Fonts](https://fonts.google.com), `.ttf` ulož vedľa notebooku a zaregistruj:
> ```python
> pdfmetrics.registerFont(TTFont('TimesNewRoman', 'Tinos-Regular.ttf'))
> ```
>
> 💡 *Tip: `Times-Roman` (možnosť 2) je zabudovaný priamo v reportlabe, takže funguje rovnako na Windows aj Macu bez akéhokoľvek súboru.*

### Nastavenie fontu v dokumente

Metóda `c.setFont()` zmení aktuálny font na zadaný typ a veľkosť:

````python
c = canvas.Canvas('created_pdf/priklad_font.pdf')

# Registrácia a nastavenie fontu
pdfmetrics.registerFont(TTFont('TimesNewRoman', 'times.ttf'))

# setFont(názov zaregistrovaného fontu, veľkosť v bodoch,
#         leading (vzdialenosť medzi riadkami pri prechode na ďalší riadok): predvolene 1.2)
c.setFont('TimesNewRoman', 35)
````

## Vloženie textu do PDF – funkcia drawString()

Na vytvorenie textového reťazca použijeme metódu `drawString()`:

```python
c.drawString(x, y, text, mode=None, charSpace=0, direction=None, wordSpace=None)
```

Parametre:
- **x, y** – súradnice začiatku reťazca. Bod x=0, y=0 je v ľavom dolnom rohu stránky
- **mode** – štýl textu (0-3): 0 = normálny, 1 = obrys, 2 = neviditeľný, 3 = obrys + výplň (tučné)
- **charSpace** – medzery medzi znakmi (celé čísla)
- **direction** – smer textu: "LTR" (zľava doprava) alebo "RTL" (sprava doľava)
- **wordSpace** – medzery medzi slovami (celé čísla)

## Funkcia showPage() – prechod na novú stránku

Funkcia `showPage()` uzavrie aktuálnu stránku. Po jej zavolaní budú všetky ďalšie kresliace príkazy vykreslené na novú stránku.

````python
width, height = A4
c = canvas.Canvas('created_pdf/viac_stranok.pdf', pagesize=A4)

c.setFont('Helvetica', 20)
c.drawString(100, height - 100, 'Text na prvej stránke')

# Prechod na novú stránku
c.showPage()

c.drawString(100, height - 100, 'Text na druhej stránke')
c.save()
````

Metóda `save()` automaticky volá `showPage()` pred uložením, ak stránka nebola uzavretá.

## Kompletný príklad – vytvorenie PDF dokumentu

````python
# Registrácia fontu
pdfmetrics.registerFont(TTFont('TimesNewRoman', 'times.ttf'))

width, height = A2
c = canvas.Canvas('created_pdf/kompletne_PDF.pdf', pagesize=A2)

# Prvá stránka
c.setFont('Helvetica', 35)
c.drawString(50, height - 50, 'Ukážkový text 1')

c.setFont('TimesNewRoman', 35)
c.drawString(150, height - 150, 'Ukážkový text 2')

# Druhá stránka - začni novú stránku a napíš nejaký text
______

c.save()
````

## Vkladanie obrázkov do PDF – funkcia `drawImage()`

Metóda `drawImage()` volaná na objekte canvas umožňuje vkladať obrázky do dokumentu:

```python
c.drawImage(image, x, y, width=None, height=None, mask=None,
            preserveAspectRatio=False, showBoundary=False)
```

Parametre:
- **image** – cesta k súboru alebo objekt ImageReader
- **x, y** – pozícia obrázka (ľavý dolný roh)
- **width, height** – šírka a výška obrázka
- **mask** – priehľadnosť (zoznam 6 čísel RGB alebo 'auto')
- **preserveAspectRatio** – zachovanie pomeru strán
- **showBoundary** – zobrazenie rámčeka

### Vloženie obrázka z URL

Na pridanie obrázka z URL musíme použiť `ImageReader`:

Dôležitá metóda triedy `ImageReader` je `getSize()`, ktorá vracia šírku a výšku stiahnutého obrázka.

````python
from reportlab.lib.utils import ImageReader
````

````python
c = canvas.Canvas('created_pdf/obrazok_priklad.pdf')
````

````python
# Stiahnutie obrázka z URL
logo = ImageReader('data/logo.png')
````

````python
# Získanie rozmerov obrázka
logo_width, logo_height = logo.getSize()
````

````python
logo_width
````

````python
logo_height
````

````python
# Vloženie obrázka (zmenšeného na polovicu)
c.drawImage(logo, 10, 10, width=logo_width / 2, height=logo_height / 2, mask='auto')
````

````python
c.save()
````

# Prehľad použitých knižníc, funkcií a metód

```python
from reportlab.pdfgen import canvas
from reportlab.lib.pagesizes import A2
from reportlab.pdfbase import pdfmetrics # pre fonty
from reportlab.pdfbase.ttfonts import TTFont # pre fonty
from reportlab.lib.utils import ImageReader # pre obrázky
from reportlab.lib.colors import lightblue # pre pozadie
```

## Základné funkcie na prácu s PDF

| Funkcia/Metóda | Použitie |
|---------------|----------|
| `canvas.Canvas(filename, pagesize)` | Vytvorenie nového PDF dokumentu |
| `A4, A3, A2, A1, A0` | Štandardné ISO formáty stránok |
| `c.save()` | Uloženie PDF dokumentu na disk |
| `c.showPage()` | Uzavretie aktuálnej stránky a prechod na novú |

## Práca s fontmi / zápis textu
| Funkcia/Metóda | Použitie |
|---------------|----------|
| `pdfmetrics.registerFont(TTFont(name, file))` | Registrácia nového fontu |
| `pdfmetrics.getRegisteredFontNames()` | Získanie zoznamu dostupných fontov |
| `TTFont(name, filename)` | Vytvorenie objektu fontu z .ttf súboru |
| `c.setFont(fontname, size)` | Nastavenie fontu a jeho veľkosti |
| `c.drawString(x, y, text)` | Vloženie textu na pozíciu [x, y] |

## Vloženie obrázka

| Funkcia/Metóda | Použitie |
|---------------|----------|
| `ImageReader(url_or_path)` | Načítanie obrázka z URL alebo cesty |
| `image.getSize()` | Získanie rozmerov obrázka (šírka, výška) |
| `c.drawImage(image, x, y, width, height)` | Vloženie obrázka na pozíciu [x, y] |

## Pozadie / farby
| Funkcia/Metóda | Použitie |
|---------------|----------|
| `c.setFillColor(farba)` | Nastaví výplňovú farbu |
| `c.rect(x1, y1, x2, y2, fill = type)` | Vyplní obdĺžnik |

## Užitočné konštanty

- **1 palec = 2,54 cm = 72 bodov**
- **Súradnice [0, 0]** = ľavý dolný roh stránky
