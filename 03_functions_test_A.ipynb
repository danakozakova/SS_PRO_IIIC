## Úloha 2: Nájdi chyby a oprav

Funkcia `priemer(a, b)` má vrátiť priemer dvoch čísel. Kód nižšie obsahuje **tri chyby**. Oprav ich tak, aby kód vypísal očakávaný výsledok.

```python
# Oprav chyby v tomto kóde:
def priemer(a, b)
    sucet = a - b
    sucet / 2

print(priemer(4, 6))   # malo by vypísať 5.0
```

## Úloha 3: Predvolený parameter, ktorý sa správa nečakane

Funkcia nižšie sa nespráva tak, ako by sme čakali (preberali sme to na hodine). Oprav ju tak, aby druhé volanie vypísalo správny výsledok.

```python
# Oprav funkciu:
def pridaj_prvok(prvok, zoznam=[]):
    zoznam.append(prvok)
    return zoznam

print(pridaj_prvok(1))   # vypíše [1]  - správne
print(pridaj_prvok(2))   # malo by vypísať [2], ale vypíše [1, 2]
```

## Úloha 4: Funkcia s predvoleným parametrom

Napíš funkciu `cena_s_dph(cena, dph=0.20)`, ktorá vráti cenu s DPH zaokrúhlenú na 2 desatinné miesta. Parameter `dph` je sadzba ako desatinné číslo a **ak sa nezadá, použije sa `0.20`**.

Príklad: `cena_s_dph(100)` vráti `120.0`, `cena_s_dph(100, 0.10)` vráti `110.0`.

```python
# Tvoje riešenie tu:
```

```python
assert cena_s_dph(100) == 120.0
assert cena_s_dph(100, 0.10) == 110.0
assert cena_s_dph(dph=0.0, cena=50) == 50.0
print('Test prešiel!')
```

## Úloha 5: Funkcia s `range` a `if`

Napíš funkciu `suma_nasobkov(do_kolko, cislo)`, ktorá vráti súčet všetkých násobkov čísla `cislo` v rozsahu od `1` do `do_kolko` (vrátane `do_kolko`). Ak `cislo` nezadáme, použije sa `2`. Ak je `cislo` nekladné, funkcia vráti `0`. Použi `range` a `if`.

```python
# Tvoje riešenie tu:
```

```python
assert suma_nasobkov(10) == 30
assert suma_nasobkov(10, 3) == 18
assert suma_nasobkov(10, 11) == 0
assert suma_nasobkov(0) == 0
assert suma_nasobkov(10, 0) == 0
assert suma_nasobkov(10, -2) == 0
assert suma_nasobkov(cislo=5, do_kolko=20) == 50
print('Test prešiel!')
```
