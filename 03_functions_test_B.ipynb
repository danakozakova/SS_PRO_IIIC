## Úloha 3: Predvolený parameter, ktorý sa správa nečakane

Funkcia nižšie sa nespráva tak, ako by sme čakali (preberali sme to na hodine). Oprav ju tak, aby druhé volanie vypísalo správny výsledok.

```python
# Oprav funkciu:
def pridaj_znamku(znamka, znamky=[]):
    znamky.append(znamka)
    return znamky

print(pridaj_znamku(1))   # vypíše [1]  - správne
print(pridaj_znamku(5))   # malo by vypísať [5], ale vypíše [1, 5]
```

## Úloha 4: Funkcia s predvoleným parametrom

Napíš funkciu `pozdrav(meno, oslovenie="Ahoj")`, ktorá vráti text v tvare `"<oslovenie>, <meno>!"`. Parameter `oslovenie` je **nepovinný** a ak sa nezadá, použije sa `"Ahoj"`.

Príklad: `pozdrav("Jana")` vráti `"Ahoj, Jana!"`, `pozdrav("Jana", "Dobrý deň")` vráti `"Dobrý deň, Jana!"`.

```python
# Tvoje riešenie tu:
```

```python
assert pozdrav("Jana") == "Ahoj, Jana!"
assert pozdrav("Jana", "Dobrý deň") == "Dobrý deň, Jana!"
assert pozdrav(oslovenie="Čau", meno="Peter") == "Čau, Peter!"
print('Test prešiel!')
```
