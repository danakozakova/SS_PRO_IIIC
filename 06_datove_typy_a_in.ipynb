## 1. Zoznam (list) &mdash; opakovanie**Zoznam** je usporiadaná kolekcia prvkov. Vlastnosti:- má **poradie** &mdash; k prvkom pristupujeme cez **index** (číslujeme od 0),- môže obsahovať **duplicity** (ten istý prvok viackrát),- je **meniteľný** &mdash; prvky vieme pridávať, meniť aj odoberať.Zoznam zapisujeme do **hranatých zátvoriek** `[ ]`.

```python
ovocie = ["jablko", "banán", "hruška"]

print(ovocie)        # celý zoznam
print(ovocie[0])     # prvý prvok (index 0)
print(ovocie[-1])    # posledný prvok
print(len(ovocie))   # počet prvkov
```

### Pridávanie, zmena a odoberanie prvkov

```python
ovocie = ["jablko", "banán", "hruška"]

ovocie.append("pomaranč")   # pridá prvok na koniec
print(ovocie)

ovocie[1] = "čerešňa"        # zmení druhý prvok
print(ovocie)

ovocie.remove("jablko")      # odoberie prvok podľa hodnoty
print(ovocie)
```

## 2. Slovník (dictionary) &mdash; opakovanie**Slovník** ukladá údaje ako páry **kľúč : hodnota**. Vlastnosti:- k hodnote pristupujeme cez **kľúč** (nie cez index),- **kľúče sú jedinečné**,- od Pythonu 3.7 si pamätá **poradie** pridania.Slovník zapisujeme do **zložených zátvoriek** `{ }`, medzi kľúč a hodnotu dávame **dvojbodku**.

```python
kniha = {
    "nazov": "Hobit",
    "autor": "J. R. R. Tolkien",
    "rok": 1937
}

print(kniha["nazov"])          # prístup k hodnote cez kľúč
kniha["pocet_stran"] = 310     # pridanie nového páru
print(kniha)
```

### Kľúče, hodnoty a páry- `keys()` &mdash; všetky kľúče,- `values()` &mdash; všetky hodnoty,- `items()` &mdash; dvojice (kľúč, hodnota).

```python
print(kniha.keys())
print(kniha.values())

# Prehľadnejšie cez for cyklus:
for kluc, hodnota in kniha.items():
    print(kluc, "->", hodnota)
```

## 3. Množina (set) &mdash; nový typ**Množina** je kolekcia, ktorá:- obsahuje **iba hodnoty** (bez kľúčov),- **neobsahuje duplicity** &mdash; každý prvok je v nej najviac raz,- **nemá poradie** &mdash; prvky nevieme indexovať (`mnozina[0]` nefunguje).Množinu zapisujeme do **zložených zátvoriek** `{ }`, podobne ako slovník, ale **bez dvojbodky** &mdash; píšeme len hodnoty.

```python
farby = {"červená", "zelená", "modrá"}
cisla = {1, 2, 3, 4, 5}

print(type(farby))
print(type(cisla))
```

Ak do množiny zapíšeme ten istý prvok viackrát, zostane v nej len **raz**:

```python
cisla = {1, 2, 2, 3, 3, 3}
print(cisla)   # {1, 2, 3}
```

### POZOR na prázdne zátvorky!Prázdne `{}` Python **vždy** považuje za slovník, nie za množinu.Prázdnu množinu vytvoríme funkciou `set()`.

```python
prazdny = {}
print(type(prazdny))   # <class 'dict'>

prazdna = set()
print(type(prazdna))   # <class 'set'>
```

### Pridávanie a odoberanie prvkov- `add()` &mdash; pridá prvok,- `discard()` &mdash; odoberie prvok (ak neexistuje, nič sa nestane),- `remove()` &mdash; odoberie prvok (ak neexistuje, vyhodí chybu),- `pop()` &mdash; odoberie „nejaký" prvok (množina nemá poradie).

```python
farby = {"červená", "zelená", "modrá"}

farby.add("žltá")        # pridá novú farbu
print(farby)

farby.add("červená")     # červená už je v množine -> nič sa nezmení
print(farby)

farby.discard("zelená")  # odoberie zelenú
print(farby)
```

## 4. Množinové operácieMnožiny podporujú operácie, ktoré poznáš z matematiky.### Prienik (intersection) &mdash; spoločné prvky &cap;Metóda `intersection()` alebo operátor `&`.

```python
mnozina1 = {1, 2, 3, 4, 5}
mnozina2 = {4, 5, 6, 7, 8}

print(mnozina1.intersection(mnozina2))   # {4, 5}
print(mnozina1 & mnozina2)               # {4, 5}
```

### Zjednotenie (union) &mdash; všetky prvky &cup;Metóda `union()` alebo operátor `|`.

```python
print(mnozina1.union(mnozina2))   # {1, 2, 3, 4, 5, 6, 7, 8}
print(mnozina1 | mnozina2)        # {1, 2, 3, 4, 5, 6, 7, 8}
```

### Rozdiel (difference) &mdash; prvky len v prvej množine &setminus;Metóda `difference()` alebo operátor `-`. Navyše `^` dáva **symetrický rozdiel** (prvky, ktoré sú len v jednej z množín).

```python
print(mnozina1.difference(mnozina2))   # {1, 2, 3}
print(mnozina1 - mnozina2)             # {1, 2, 3}

print(mnozina1 ^ mnozina2)             # {1, 2, 3, 6, 7, 8}
```

### Praktické využitie: odstránenie duplicítNajčastejšie využitie množiny &mdash; rýchlo **odstrániť duplicity** zo zoznamu.

```python
cisla = [1, 2, 2, 3, 3, 3, 4]
bez_duplicit = set(cisla)
print(bez_duplicit)         # {1, 2, 3, 4}

print(list(bez_duplicit))   # a späť na zoznam
```

### Množina vs slovník &mdash; nezameň ich!| | zápis | obsah ||---|---|---|| **množina** | `{"a", "b", "c"}` | iba hodnoty || **slovník** | `{"a": 1, "b": 2}` | páry kľúč : hodnota (s dvojbodkou) |

## 5. Kľúčové slovo `in`Kľúčové slovo `in` je v Pythone veľmi univerzálne. Používame ho na:- **iterovanie** &mdash; prechádzanie prvkov kolekcie (`for prvok in kolekcia`),- **kontrolu prítomnosti** &mdash; či sa prvok nachádza v kolekcii (vráti `True` / `False`),- **vyhľadávanie v texte** &mdash; či je jeden reťazec časťou druhého.

### 5.1 Iterovanie cez zoznamCyklus `for` s `in` prejde postupne všetky prvky zoznamu.

```python
zvierata = ["mačka", "pes", "myš"]

for zviera in zvierata:
    print(zviera)
```

### 5.2 Iterovanie cez slovníkPri prechádzaní slovníka dostávame **kľúče**. K hodnote sa dostaneme cez `slovnik[kluc]`.

```python
kniha = {"nazov": "Hobit", "autor": "J. R. R. Tolkien", "rok": 1937}

for kluc in kniha:
    print(kluc, ":", kniha[kluc])
```

### 5.3 Iterovanie cez množinu a reťazecCez `for ... in` vieme prejsť aj množinu (poradie nie je zaručené) a dokonca aj znaky reťazca.

```python
for farba in {"červená", "zelená", "modrá"}:
    print(farba)

print("---")

for znak in "Python":
    print(znak)
```

### 5.4 Kontrola prítomnosti v zozname`in` vráti `True`, ak je prvok v kolekcii, inak `False`.

```python
zvierata = ["mačka", "pes", "myš"]

print("pes" in zvierata)       # True
print("sliepka" in zvierata)   # False
```

Najčastejšie `in` spojíme s podmienkou `if`:

```python
ovocie = ["jablko", "hruška", "jahoda", "melón"]

if "jahoda" in ovocie:
    print("Jahoda je v zozname")
else:
    print("Jahoda nie je v zozname")
```

### 5.5 Kontrola prítomnosti v množine (rýchla!)V množine je kontrola `in` **oveľa rýchlejšia** než v zozname. Ak často zisťuješ, či prvok existuje, oplatí sa použiť množinu.

```python
velke_cisla = set(range(1_000_000))

print(999_999 in velke_cisla)   # True — bleskovo rýchle
```

### 5.6 Kontrola kľúča v slovníkuV slovníku `in` kontroluje, či existuje daný **kľúč** (nie hodnota!).

```python
zviera = {"name": "cat", "breed": "moggy", "owner": "Baba Yaga"}

print("breed" in zviera)   # True  — "breed" je kľúč
print("cat" in zviera)     # False — "cat" je hodnota, nie kľúč
```

```python
student = {"meno": "Jan", "trieda": "3.A"}

if "vek" in student:
    print("Vek:", student["vek"])
else:
    print("Vek nie je uvedený")
```

### 5.7 Vyhľadávanie v texte`in` zistí, či je jeden reťazec obsiahnutý v druhom. **Pozor na veľké a malé písmená** &mdash; `"Python"` a `"python"` nie je to isté.

```python
veta = "The quick brown fox jumps over the lazy dog"

print("fox" in veta)   # True
print("cat" in veta)   # False
print("Fox" in veta)   # False — veľké F
```

## 6. Kombinované štruktúryKolekcie môžeme do seba **vnárať** &mdash; slovník môže mať ako hodnotu zoznam, zoznam môže obsahovať slovníky atď.

### 6.1 Zoznam slovníkovTypické pre tabuľkové údaje &mdash; každý slovník je jeden „riadok".

```python
studenti = [
    {"meno": "Anna", "vek": 20, "znamka": 1},
    {"meno": "Peter", "vek": 22, "znamka": 2},
]

print(studenti[0]["meno"])   # meno prvého študenta -> Anna

for student in studenti:
    print(student["meno"], "má známku", student["znamka"])
```

### 6.2 Slovník zoznamovNapríklad trieda a zoznam jej žiakov.

```python
skola = {
    "1.A": ["Anna", "Pavol", "Jana"],
    "1.B": ["Peter", "Mária", "Ivan"],
}

print(skola["1.A"])        # celá trieda 1.A
print(skola["1.A"][1])     # druhý žiak v 1.A -> Pavol

skola["1.B"].append("Eva") # pridáme žiačku do 1.B
print(skola["1.B"])
```

### 6.3 Prepojenie: `in`, množiny a dátaUkážka, ako témy fungujú spolu &mdash; zistíme, ktoré jazyky majú dvaja študenti spoločné.

```python
jazyky_jana = {"Python", "Java", "C++", "JavaScript"}
jazyky_peter = {"Python", "Ruby", "JavaScript", "Go"}

spolocne = jazyky_jana & jazyky_peter
print("Spoločné:", spolocne)

if "Python" in spolocne:
    print("Obaja vedia Python")
```
