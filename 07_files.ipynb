## 6. Absolútne cesty

Absolútna cesta je taká, ktorá jednoznačne určuje, kde sa konkrétny súbor nachádza na disku - nezáleží na tom, z ktorého priečinka bol program spustený.

Príklad absolútnej cesty v macOS/Linux: `/home/jconnor/Documents/moj_subor.txt`

Príklad absolútnej cesty vo Windows: `C:\Documents and Settings\John Connor\Documents\moj_subor.txt`

Rozhodujúcim prvkom je tu znak `/` alebo `C:\`.

**Odporúčanie:** Pri písaní skriptov používajte relatívne cesty (takýto kód sa lepšie prenáša medzi používateľmi a dokonca aj systémami); len si buďte vedomí dôsledkov spúšťania skriptov z rôznych umiestnení na disku.

## 7. Cesty vo Windows

Systémy Windows používajú na oddeľovanie názvov súborov a priečinkov znak `\`.

To môže spôsobiť problémy: čítanie súboru `data\chart.jpg` uspeje, zatiaľ čo `data\toto.jpg` nie.

Je to preto, že niektoré postupnosti znakov (`\a`, `\b`, `\n`, `\r`, `\t`) majú špeciálny význam - v uvedenom príklade: `\t` bol nahradený znakom tabulátora, čím stratil svoj zamýšľaný význam oddeliť názov priečinka od názvu súboru.

````python
print('data\toto.jpg')
````

### 7.1. Riešenie 1: Použite `/`

````python
# Použite znak / a nechajte Python ho v prípade potreby automaticky nahradiť za \
with open('data/toto.jpg') as subor:
    pass
````

### 7.2. Riešenie 2: Raw string

````python
# Zadajte cesty pomocou syntaxe raw string: r'relativna\cesta\k\suboru.txt'
with open(r'data\toto.jpg') as subor:
    pass
````

### 7.3. Riešenie 3: Zdvojenie spätného lomítka

````python
# Pred znak \ dajte druhý znak \: 'relativna\\cesta\\k\\suboru.txt'
with open('data\\toto.jpg') as subor:
    pass
````

## 9. Vyhľadávanie súborov v priečinkoch

Najjednoduchší spôsob, ako hľadať súbory v priečinku, je funkcia `glob` z modulu `glob`.

Stačí jej odovzdať vzor pre názov (alebo cestu) súborov, ktoré chcete nájsť; znak `*` znamená, že na tomto mieste v názve môže byť čokoľvek.

````python
os.getcwd()
````

````python
os.chdir('data')
````

````python
import glob

# Nájde všetky súbory s príponou .txt
glob.glob('*.txt')  # ['moj_subor.txt', 'moj_dalsi_subor.txt', 'test.txt']
````

````python
# Nájde všetky súbory začínajúce na 'moj_'
glob.glob('moj_*.txt')  # ['moj_subor.txt', 'moj_dalsi_subor.txt']
````

### 9.1. Rekurzívne vyhľadávanie

````python
# Funkcia môže prijať aj ďalší parameter recursive=True,
# takže začne hľadať aj vnútri priečinkov.
# Znaky ** potom označujú, kde v ceste môžu byť priečinky s neznámymi názvami
import glob

# Hľadá súbory .jpg vo všetkých podpriečinkoch
glob.glob(r'**\*.jpg', recursive=True)  # ['jun/praca/spanielsko.jpg', 'august/praca/taliansko.jpg']
````

````python
# Hľadá súbory .jpg iba v priečinku jun a jeho podpriečinkoch
glob.glob(r'jun\**\*.jpg', recursive=True)  # ['jun/praca/spanielsko.jpg']
````
