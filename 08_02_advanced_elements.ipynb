> 💡 **Slovenská diakritika v `Paragraph`:** Predvolený font `Helvetica` nepodporuje slovenské znaky (`č, š, ž, ť, ľ, ň…`). Preto zaregistrujeme celú rodinu `Arial` (vrátane tučného a kurzívy cez `registerFontFamily`, aby fungovali aj tagy `<b>` a `<i>`) a nastavíme ju v `styles`.
>
> *Na macOS zadajte cestu `/System/Library/Fonts/Supplemental/Arial.ttf` (a `Arial Bold.ttf`, `Arial Italic.ttf`, `Arial Bold Italic.ttf`) – pozri poznámku v `08_01`.*

````python
from reportlab.pdfbase import pdfmetrics
from reportlab.pdfbase.ttfonts import TTFont
from reportlab.pdfbase.pdfmetrics import registerFontFamily

# Registrácia rodiny Arial (na Windows sa nájde automaticky v C:\Windows\Fonts)
pdfmetrics.registerFont(TTFont('Arial', 'arial.ttf'))
pdfmetrics.registerFont(TTFont('Arial-Bold', 'arialbd.ttf'))
pdfmetrics.registerFont(TTFont('Arial-Italic', 'ariali.ttf'))
pdfmetrics.registerFont(TTFont('Arial-BoldItalic', 'arialbi.ttf'))
registerFontFamily(
    'Arial',
    normal='Arial',
    bold='Arial-Bold',
    italic='Arial-Italic',
    boldItalic='Arial-BoldItalic',
)

def nastav_slovensky_font(styles):
    for s in styles.byName.values():
        if 'Bold' in s.fontName:
            s.fontName = 'Arial-Bold'
        elif 'Oblique' in s.fontName or 'Italic' in s.fontName:
            s.fontName = 'Arial-Italic'
        elif s.fontName == 'Helvetica':
            s.fontName = 'Arial'
    return styles

styles = nastav_slovensky_font(styles)
````

### Paragraph – odsek

Trieda `Paragraph` umožňuje umiestniť jeden odsek textu. Dá sa štýlovať dvoma spôsobmi:
1. Pomocou XML tagov v texte
2. Pomocou štýlu `ParagraphStyle`

````python
# Základná syntax Paragraph
# Paragraph(text, style, bulletText=None)

styles = nastav_slovensky_font(getSampleStyleSheet())
p = Paragraph("Toto je ukážkový odsek.", styles["Normal"])
````

#### Úprava textu v odseku pomocou XML tagov

Pomocou XML tagov môžeme ovplyvniť vzhľad textu:

- `<para>...</para>` – štýlovanie celého odseku
- `<font>...</font>` – zmena písma pre časť textu
- `<b>...</b>` – tučný text
- `<i>...</i>` – kurzíva
- `<u>...</u>` – podčiarknutie
- `<super>...</super>` a `<sub>...</sub>` – horný a dolný index

#### `<para>...</para>`

Tag `<para>` musí obklopiť celý text odseku. Pomocou jeho atribútov nastavujeme štýl celému odseku.

````python
doc = SimpleDocTemplate("created_pdf/para_tag_example.pdf", pagesize=A4,
                        rightMargin=72, leftMargin=72,
                        topMargin=72, bottomMargin=18)
````

````python
story = []
````

````python
# Text s tagom <para> - nastavenie okrajov, farby pozadia a zarovnania
ptext = '<para borderWidth="1" backColor="yellow" borderColor="black" alignment="right">Lorem ipsum dolor sit amet...</para>'

story.append(Paragraph(ptext, styles["Normal"]))
````

````python
doc.build(story)
````

#### `<font>...</font>`

Tag `<font>` nastavuje štýl pre text, ktorý obsahuje – písmo, farbu, veľkosť.

````python
doc = SimpleDocTemplate("created_pdf/font_tag_example.pdf", pagesize=A4,
                        rightMargin=72, leftMargin=72,
                        topMargin=72, bottomMargin=18)
````

````python
story = []
````

````python
# Kombinácia <para> a <font>
ptext = '<para borderWidth="1">Lorem ipsum dolor sit amet, <font face="times" color="red" size="16">consectetur adipiscing elit</font>, sed do eiusmod tempor.</para>'

story.append(Paragraph(ptext, styles["Normal"]))
````

````python
doc.build(story)
````

#### Ostatné tagy – b, i, u, super, sub

````python
doc = SimpleDocTemplate("created_pdf/other_tags_example.pdf", pagesize=A4,
                        rightMargin=72, leftMargin=72,
                        topMargin=72, bottomMargin=18)
````

````python
story = []
````

````python
# Ukážka rôznych tagov
ptext = '<u>Podčiarknutý text</u>, <b>tučný text</b>, <i>kurzíva</i>'
story.append(Paragraph(ptext, styles["Normal"]))

story.append(Spacer(1, 12))

# Horný a dolný index
ptext = 'H<sub>2</sub>O je vzorec vody, E=mc<super>2</super> je slávna rovnica.'
story.append(Paragraph(ptext, styles["Normal"]))
````

````python
doc.build(story)
````

#### Štýlovanie celého odseku – ParagraphStyle

Celý odsek sa dá štýlovať pomocou triedy `ParagraphStyle`. Knižnica poskytuje základné štýly pomocou `getSampleStyleSheet()`:

- `Normal` – normálny, základný text
- `Italic` – kurzíva
- `Heading1` až `Heading6` – rôzne štýly nadpisov
- `Title` – štýl pre titulok dokumentu

````python
doc = SimpleDocTemplate("created_pdf/styles_example.pdf", pagesize=A4,
                        rightMargin=72, leftMargin=72,
                        topMargin=72, bottomMargin=18)
styles = nastav_slovensky_font(getSampleStyleSheet())
story = []

# Použitie rôznych štýlov
story.append(Paragraph("Toto je Title", styles["Title"]))
story.append(Spacer(1, 12))
story.append(Paragraph("Toto je Heading1", styles["Heading1"]))
story.append(Spacer(1, 12))
story.append(Paragraph("Toto je Normal text.", styles["Normal"]))
story.append(Spacer(1, 12))
story.append(Paragraph("Toto je Italic text.", styles["Italic"]))

doc.build(story)
print("Súbor styles_example.pdf bol vytvorený.")
````

#### Vytvorenie vlastného štýlu

````python
from reportlab.lib.styles import ParagraphStyle

doc = SimpleDocTemplate("created_pdf/custom_style_example.pdf", pagesize=A4,
                        rightMargin=72, leftMargin=72,
                        topMargin=72, bottomMargin=18)
styles = nastav_slovensky_font(getSampleStyleSheet())
story = []

# Vytvorenie vlastného štýlu
bigStyle = ParagraphStyle('big',
                          fontName="Arial-Bold",
                          fontSize=16,
                          parent=styles['Normal'])

ptext = 'Lorem ipsum dolor sit amet, consectetur adipiscing elit.'
story.append(Paragraph(ptext, bigStyle))

doc.build(story)
print("Súbor custom_style_example.pdf bol vytvorený.")
````

### Image – obrázok

Trieda `Image` pridáva obrázok do dokumentu.

**Parametre:**
- `filename` – cesta alebo URL k obrázku
- `width` – šírka v bodoch (pt) (voliteľné)
- `height` – výška v bodoch (pt) (voliteľné)

Ak `width` a `height` neurčíme, obrázok si zachová pôvodné rozmery.

````python
from reportlab.platypus import Image

doc = SimpleDocTemplate("created_pdf/image_example.pdf", pagesize=A4,
                        rightMargin=72, leftMargin=72,
                        topMargin=72, bottomMargin=18)
story = []
````

````python
# Pridanie obrázka s pôvodnou veľkosťou
im = Image('data/logo.png')
story.append(im)
story.append(Spacer(1, 20))
````

````python
# Pridanie obrázka so zmenenou veľkosťou
im_resized = Image('data/logo.png', 160, 56)
story.append(im_resized)
````

````python
doc.build(story)
````
