## 5. Funkcia s dvoma parametrami

Táto funkcia prijíma dva parametre a vracia čistú sumu vypočítanú na ich základe.

## 6. Funkcie s predvolenými hodnotami parametrov

## 9. Telo funkcie - zložitejšia logika

Predchádzajúce príklady boli veľmi jednoduché, pretože zahŕňali iba jeden základný výpočet.

Prakticky neexistujú žiadne obmedzenia, čo môžeme vložiť do tela funkcie. Cykly, podmienené príkazy atď. - všetko je dovolené!

````python
def zlozeny_urok(suma, urok, obdobie):
    """Vypočíta hodnotu investície po zadanom počte období so zloženým úrokom."""
    for p in range(obdobie):
        suma = suma * (1 + urok)
    return round(suma, 2)

# Použitie funkcie
vysledok = zlozeny_urok(1000, 0.02, 5)
print(f'1000 EUR po 5 rokoch na 2 % vklade zarobí {vysledok} EUR')
````
