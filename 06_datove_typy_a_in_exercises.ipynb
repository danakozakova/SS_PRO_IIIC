Ak sa `slovo` nachádza v `zoznam_slov`, pridá na koniec zoznamu reťazec `"Found"`. Ak sa tam nenachádza, zoznam **nemení**.

Správnosť si over spustením testovacích buniek nižšie.

```python
# Napíš funkciu wordfinder sem
```

```python
# Test 1 — slovo sa v zozname nachádza
slova = ['Cats', 'coffee', 'cheesecake', 'unite']
wordfinder(slova, 'coffee')
if slova == ['Cats', 'coffee', 'cheesecake', 'unite', 'Found']:
    print('Test 1 OK!')
else:
    print('Chyba: funkcia nepridala "Found" na koniec zoznamu.')
```

```python
# Test 2 — slovo sa v zozname nenachádza
slova2 = ['Delights', 'to', 'cherish', 'hearts', 'unbound']
wordfinder(slova2, 'coffee')
if slova2 == ['Delights', 'to', 'cherish', 'hearts', 'unbound']:
    print('Test 2 OK!')
else:
    print('Chyba: funkcia zbytočne zmenila zoznam.')
```
