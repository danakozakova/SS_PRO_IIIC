## Cvičenie 1: Používateľské údaje

Napíš funkciu `pouzivatelske_data(meno, priezvisko)`, ktorá prijíma ako argumenty meno a priezvisko a vracia slovník s nasledujúcimi kľúčmi a hodnotami:
- `"meno"` – hodnota z argumentu meno,
- `"priezvisko"` – hodnota z argumentu priezvisko,
- `"cele_meno"` – zložené z oboch argumentov oddelených medzerou,
- `"inicialy"` – prvé písmená oboch argumentov nasledované bodkami; obe časti oddelené medzerou.

**Príklad:**
```python
pouzivatelske_data('Jan', 'Novák')
```

**Výsledok:**
```python
{"meno": "Jan", "priezvisko": "Novák", "cele_meno": "Jan Novák", "inicialy": "J. N."}
```

````python
# Sem napíš funkciu "pouzivatelske_data"
````

### Testovanie funkcie

````python
if pouzivatelske_data('Jan', 'Novák') == {
        "meno": "Jan",
        "priezvisko": "Novák",
        "cele_meno": "Jan Novák",
        "inicialy": "J. N."
}:
    print(' Test prešiel!')
else:
    print('Funkcia nevrátila očakávaný výsledok. Očakávané:')
    print('{"meno": "Jan", "priezvisko": "Novák", "cele_meno": "Jan Novák", "inicialy": "J. N."}')
    print('Funkcia vrátila:')
    print(pouzivatelske_data('Jan', 'Novák'))

if pouzivatelske_data('Anna', 'Svobodová') == {
        "meno": "Anna",
        "priezvisko": "Svobodová",
        "cele_meno": "Anna Svobodová",
        "inicialy": "A. S."
}:
    print(' Test prešiel!')
else:
    print(' Funkcia nevrátila očakávaný výsledok. Očakávané:')
    print('{"meno": "Anna", "priezvisko": "Svobodová", "cele_meno": "Anna Svobodová", "inicialy": "A. S."}')
    print('Funkcia vrátila:')
    print(pouzivatelske_data('Anna', 'Svobodová'))
````

## Cvičenie 2: Zobrazenie slovníka

Napíš funkciu `zobraz_slovnik(slovnik)`, ktorá prijíma slovník ako argument a zobrazí ho spôsobom uvedeným nižšie.

**Príklad:**
```python
zobraz_slovnik({'macka': 'Mačka', 'pes': 'Pes', 'vtak': 'Vták'})
```

**Výstup:**
```
macka=>Mačka
pes=>Pes
vtak=>Vták
```

**Úloha:** Použi cyklus `for` na prechádzanie slovníka a vypíš každú dvojicu kľúč-hodnota na nový riadok.

````python
# Sem napíš funkciu "zobraz_slovnik"
````

### Testovanie funkcie

````python
import sys
import io
_real_stdout = sys.stdout

sys.stdout = io.StringIO()
zobraz_slovnik({'macka': 'Mačka', 'pes': 'Pes'})
sys.stdout.seek(0)
if set(sys.stdout.read().strip().split()) == {'macka=>Mačka', 'pes=>Pes'}:
    print(' Test prešiel!', file=_real_stdout)
else:
    print(' Funkcia nezobrazila očakávané riadky', file=_real_stdout)
    print('Funkcia zobrazila:', file=_real_stdout)
    sys.stdout.seek(0)
    print(sys.stdout.read(), file=_real_stdout)
    print('Namiesto očakávaného:\nmacka=>Mačka\npes=>Pes\n', file=_real_stdout)

sys.stdout = io.StringIO()
zobraz_slovnik({'vtak': 'Vták', 'mys': 'Myš'})
sys.stdout.seek(0)
if set(sys.stdout.read().strip().split()) == {'vtak=>Vták', 'mys=>Myš'}:
    print(' Test prešiel!', file=_real_stdout)
else:
    print(' Funkcia nezobrazila očakávané riadky', file=_real_stdout)
    print('Funkcia zobrazila:', file=_real_stdout)
    sys.stdout.seek(0)
    print(sys.stdout.read(), file=_real_stdout)
    print('Namiesto očakávaného:\nvtak=>Vták\nmys=>Myš\n', file=_real_stdout)

sys.stdout = _real_stdout
````

## Cvičenie 3: Bezpečný prístup ku kľúču

Napíš funkciu `bezpecny_pristup(slovnik, kluc)`, ktorá vráti hodnotu zadaného kľúča.
Ak kľúč v slovníku neexistuje, vráti reťazec `"nenájdené"` (namiesto chyby `KeyError`).

**Tip:** použi metódu `.get()`.

**Príklad:**
```python
bezpecny_pristup({"meno": "Jana", "vek": 20}, "vek")
bezpecny_pristup({"meno": "Jana", "vek": 20}, "mesto")
```

**Výsledok:**
```python
20
'nenájdené'
```

````python
# Sem napíš funkciu "bezpecny_pristup"
````

### Testovanie funkcie

````python
if (bezpecny_pristup({"meno": "Jana", "vek": 20}, "vek") == 20
        and bezpecny_pristup({"meno": "Jana", "vek": 20}, "mesto") == "nenájdené"):
    print(" Test prešiel!")
else:
    print(" Test NEPREŠIEL — skontroluj funkciu.")
    print("existujúci kľúč:", bezpecny_pristup({"meno": "Jana", "vek": 20}, "vek"))
    print("neexistujúci kľúč:", bezpecny_pristup({"meno": "Jana", "vek": 20}, "mesto"))
````

## Cvičenie 4: Súčet hodnôt

Napíš funkciu `suma_hodnot(slovnik)`, ktorá vráti súčet všetkých (číselných) hodnôt v slovníku.
Pre prázdny slovník `{}` vráti `0`.

**Tip:** prejdi hodnoty cez `.values()`.

**Príklad:**
```python
suma_hodnot({"chlieb": 10, "maslo": 5, "mlieko": 3})
```

**Výsledok:**
```python
18
```

````python
# Sem napíš funkciu "suma_hodnot"
````

### Testovanie funkcie

````python
if suma_hodnot({"chlieb": 10, "maslo": 5, "mlieko": 3}) == 18 and suma_hodnot({}) == 0:
    print(" Test prešiel!")
else:
    print(" Test NEPREŠIEL — skontroluj funkciu.")
    print("Tvoj výsledok:", suma_hodnot({"chlieb": 10, "maslo": 5, "mlieko": 3}))
````

## Cvičenie 5: Najdrahšia položka

Slovník `ceny` obsahuje dvojice `názov položky : cena`. Napíš funkciu `najdrahsi(ceny)`,
ktorá vráti **názov** položky s najvyššou cenou.

**Tip:** prejdi dvojice cez `.items()` a priebežne si pamätaj najdrahšiu.

**Príklad:**
```python
najdrahsi({"chlieb": 1.2, "maslo": 2.5, "mlieko": 0.9})
```

**Výsledok:**
```python
'maslo'
```

````python
# Sem napíš funkciu "najdrahsi"
````

### Testovanie funkcie

````python
if (najdrahsi({"chlieb": 1.2, "maslo": 2.5, "mlieko": 0.9}) == "maslo"
        and najdrahsi({"jediny": 1}) == "jediny"):
    print(" Test prešiel!")
else:
    print(" Test NEPREŠIEL — skontroluj funkciu.")
    print("Tvoj výsledok:", najdrahsi({"chlieb": 1.2, "maslo": 2.5, "mlieko": 0.9}))
````

## Cvičenie 6: Frekvencia písmen

Napíš funkciu `frekvencia_pismen(text)`, ktorá vráti slovník, kde ku každému písmenu priradí,

**Výsledok:**

````python
# Sem napíš funkciu "frekvencia_pismen"
````

### Testovanie funkcie

````python
if frekvencia_pismen("ahoj Anna") == {"a": 3, "h": 1, "o": 1, "j": 1, "n": 2}:
    print(" Test prešiel!")
else:
    print(" Test NEPREŠIEL — skontroluj funkciu.")
    print("Tvoj výsledok:", frekvencia_pismen("ahoj Anna"))
````

## Cvičenie 7: Zlúčenie slovníkov so sčítaním

Napíš funkciu `zluc_a_scitaj(a, b)`, ktorá vráti **nový** slovník vzniknutý zlúčením slovníkov `a` a `b`.
Ak sa kľúč nachádza v oboch, jeho hodnoty **spočítaj**. Pôvodné slovníky nemeň.

**Tip:** začni kópiou prvého slovníka: `vysledok = dict(a)`.

**Príklad:**
```python
zluc_a_scitaj({"jablká": 3, "hrušky": 2}, {"hrušky": 5, "slivky": 1})
```

**Výsledok:**
```python
{'jablká': 3, 'hrušky': 7, 'slivky': 1}
```

````python
# Sem napíš funkciu "zluc_a_scitaj"
````

### Testovanie funkcie

````python
a = {"jablká": 3, "hrušky": 2}
b = {"hrušky": 5, "slivky": 1}
vysl = zluc_a_scitaj(a, b)
if vysl == {"jablká": 3, "hrušky": 7, "slivky": 1} and a == {"jablká": 3, "hrušky": 2}:
    print(" Test prešiel!")
else:
    print(" Test NEPREŠIEL — skontroluj funkciu (a nemeň pôvodné slovníky).")
    print("Tvoj výsledok:", vysl)
````

## Cvičenie 8: Priemer známok (vnorený slovník)

Vstupom je slovník, kde ku každému menu je priradený **zoznam známok**, napr.
`{"Jana": [1, 2, 1], "Peter": [3, 3, 4, 2]}`.
Napíš funkciu `priemer_znamok(studenti)`, ktorá vráti slovník `{meno: priemer}`,
pričom priemer zaokrúhliš na 2 desatinné miesta pomocou `round(..., 2)`.

**Tip:** `sum(znamky) / len(znamky)`.

**Príklad:**
```python
priemer_znamok({"Jana": [1, 2, 1], "Peter": [3, 3, 4, 2]})
```

**Výsledok:**
```python
{'Jana': 1.33, 'Peter': 3.0}
```

````python
# Sem napíš funkciu "priemer_znamok"
````

### Testovanie funkcie

````python
if priemer_znamok({"Jana": [1, 2, 1], "Peter": [3, 3, 4, 2]}) == {"Jana": 1.33, "Peter": 3.0}:
    print(" Test prešiel!")
else:
    print(" Test NEPREŠIEL — skontroluj funkciu.")
    print("Tvoj výsledok:", priemer_znamok({"Jana": [1, 2, 1], "Peter": [3, 3, 4, 2]}))
````
