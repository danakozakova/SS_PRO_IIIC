## A) Kódy — ako počítač pozná preklep

Každý **čiarový kód (EAN-13)**, **IBAN** či **rodné číslo** má navyše **kontrolnú číslicu**, ktorá sa dopočíta z ostatných. Keď sa niekto pomýli, kontrola nesedí. Srdcom je **modulo**.


### EAN-13 (čiarový kód)
Z prvých 12 cifier: cifry na nepárnych pozíciách ×1, na párnych ×3, sčítaj, a 13. (kontrolná) číslica je `(10 − súčet mod 10) mod 10`. Spusti demo:


````python
kod = "4006381333931"
cifry = [int(z) for z in kod]
sucet = sum(cifry[i] * (1 if i % 2 == 0 else 3) for i in range(12))
kontrolna = (10 - sucet % 10) % 10
print("súčet:", sucet, "→ kontrolná:", kontrolna, "→ v kóde je:", cifry[12])
````

**❓** Zober si reálny produkt zo stola a over jeho čiarový kód (13 cifier).

## B) Farby — RGB a hex

Farba na webe je `#RRGGBB` — tri zložky (červená, zelená, modrá), každá 0–255, zapísané **šestnástkovo**. Prevody sú len sústavy a konverzie.


````python
h = "#3498DB"
r = int(h[1:3], 16)
g = int(h[3:5], 16)
b = int(h[5:7], 16)
print(r, g, b)
print(f"späť: #{r:02X}{g:02X}{b:02X}")
````

Vieš farbu aj **vidieť** — spusti (funguje v Colabe aj Jupyteri):


````python
from IPython.display import HTML
HTML(f'<div style="width:80px;height:40px;background:{h}"></div>')
````

### Bitový bonus
Celú farbu sa dá „zbaliť" do jedného čísla: `(r<<16) | (g<<8) | b`, a rozbaliť posunmi a maskou `& 0xFF`. Spusti:


````python
cislo = (r << 16) | (g << 8) | b
print(cislo, hex(cislo))
print((cislo >> 16) & 0xFF, (cislo >> 8) & 0xFF, cislo & 0xFF)
````

## C) Peniaze — prečo nie `float`

Skús spočítať tri desaťcentovky vo `float`:


````python
print(0.10 + 0.10 + 0.10)        # 0.30000000000000004
print(0.10 + 0.10 + 0.10 == 0.30)  # False
````

Vo `float` sa ceny **nedajú** počítať presne — na peniaze sa používa `Decimal`. Spusti:


````python
from decimal import Decimal, ROUND_HALF_UP
zaklad = Decimal('12.50')
sdph = (zaklad * Decimal('1.20')).quantize(Decimal('0.01'), rounding=ROUND_HALF_UP)
print("s DPH 20%:", sdph, "€")
````
